```python
import os
import cv2
import time
import torch
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import seaborn as sns
import torchvision
import warnings
warnings.filterwarnings('ignore')

from ultralytics import YOLO
from PIL import Image
from torchvision.ops import box_iou

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f" Device: {device}")
```

```python
image_folder = '/content/coco128/coco128/images/train2017'
label_folder = '/content/coco128/coco128/labels/train2017'

all_images = [f for f in os.listdir(image_folder)
              if f.endswith('.jpg') or f.endswith('.png')]
print(f" Total images: {len(all_images)}")

test_images = all_images[:30]
print(f"Using {len(test_images)} images")
```

```python
print("Loading YOLOv8m...")
yolo_model = YOLO('yolov8m.pt')
print(" YOLO loaded!")

print("Loading Faster R-CNN...")
frcnn_model = torchvision.models.detection.fasterrcnn_resnet50_fpn(
    pretrained=True
)
frcnn_model.eval()
frcnn_model = frcnn_model.to(device)
print("Faster R-CNN loaded!")
print("All models ready!")
```

```python
def preprocess_for_frcnn(img_path):
    image = cv2.imread(img_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    tensor = torch.tensor(
        image_rgb / 255.0, dtype=torch.float32
    ).permute(2, 0, 1).unsqueeze(0).to(device)
    return tensor, image_rgb

def compute_iou(box1, box2):
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])
    inter = max(0, x2 - x1) * max(0, y2 - y1)
    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])
    union = area1 + area2 - inter
    return inter / union if union > 0 else 0

print(" Helper functions ready!")
```

```python
print("▶ Running Standalone YOLO...")

yolo_times = []
yolo_boxes_all = []
yolo_scores_all = []
yolo_labels_all = []
yolo_det_counts = []

for img_name in test_images:
    img_path = os.path.join(image_folder, img_name)

    start = time.time()
    results = yolo_model(img_path, conf=0.5, verbose=False)
    end = time.time()
    yolo_times.append(end - start)

    boxes = results[0].boxes
    if boxes is not None and len(boxes) > 0:
        yolo_boxes_all.append(boxes.xyxy.cpu().numpy())
        yolo_scores_all.append(boxes.conf.cpu().numpy())
        yolo_labels_all.append(boxes.cls.cpu().numpy())
        yolo_det_counts.append(len(boxes))
    else:
        yolo_boxes_all.append(np.array([]))
        yolo_scores_all.append(np.array([]))
        yolo_labels_all.append(np.array([]))
        yolo_det_counts.append(0)

avg_yolo_time = np.mean(yolo_times)
yolo_fps = 1 / avg_yolo_time

print(f" YOLO Done!")
print(f"   Avg Time : {avg_yolo_time:.3f}s")
print(f"   FPS      : {yolo_fps:.1f}")
print(f"   Total Detections: {sum(yolo_det_counts)}")
```

```python
print("▶ Running Standalone Faster R-CNN...")

frcnn_times = []
frcnn_boxes_all = []
frcnn_scores_all = []
frcnn_labels_all = []
frcnn_det_counts = []

for img_name in test_images:
    img_path = os.path.join(image_folder, img_name)
    tensor, _ = preprocess_for_frcnn(img_path)

    start = time.time()
    with torch.no_grad():
        output = frcnn_model(tensor)
    end = time.time()
    frcnn_times.append(end - start)

    scores = output[0]['scores'].cpu().numpy()
    keep = scores > 0.5

    kept_boxes  = output[0]['boxes'].cpu().numpy()[keep]
    kept_scores = scores[keep]
    kept_labels = output[0]['labels'].cpu().numpy()[keep]

    frcnn_boxes_all.append(kept_boxes)
    frcnn_scores_all.append(kept_scores)
    frcnn_labels_all.append(kept_labels)
    frcnn_det_counts.append(len(kept_boxes))

avg_frcnn_time = np.mean(frcnn_times)
frcnn_fps = 1 / avg_frcnn_time

print(f"Faster R-CNN Done!")
print(f"   Avg Time : {avg_frcnn_time:.3f}s")
print(f"   FPS      : {frcnn_fps:.1f}")
print(f"   Total Detections: {sum(frcnn_det_counts)}")
```

```python
print("▶ Running Hybrid YOLO + Faster R-CNN...")

hybrid_times = []
hybrid_boxes_all = []
hybrid_scores_all = []
hybrid_labels_all = []
hybrid_det_counts = []

for img_name in test_images:
    img_path = os.path.join(image_folder, img_name)

    start = time.time()

    # Stage 1 — YOLO proposals
    yolo_out = yolo_model(img_path, conf=0.5, verbose=False)
    yolo_boxes = yolo_out[0].boxes

    image = cv2.imread(img_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    h, w = image_rgb.shape[:2]

    final_boxes = []
    final_scores = []
    final_labels = []

    if yolo_boxes is not None and len(yolo_boxes) > 0:
        proposals = yolo_boxes.xyxy.cpu().numpy()

        # Stage 2 — Faster R-CNN on each crop
        for box in proposals:
            x1, y1, x2, y2 = map(int, box)
            x1, y1 = max(0, x1), max(0, y1)
            x2, y2 = min(w, x2), min(h, y2)

            if x2 - x1 < 10 or y2 - y1 < 10:
                continue

            crop = image_rgb[y1:y2, x1:x2]
            crop_resized = cv2.resize(crop, (224, 224))

            tensor = torch.tensor(
                crop_resized / 255.0, dtype=torch.float32
            ).permute(2, 0, 1).unsqueeze(0).to(device)

            with torch.no_grad():
                out = frcnn_model(tensor)

            scores = out[0]['scores'].cpu().numpy()
            if len(scores) > 0 and scores[0] > 0.5:
                final_boxes.append([x1, y1, x2, y2])
                final_scores.append(float(scores[0]))
                final_labels.append(
                    int(out[0]['labels'].cpu().numpy()[0])
                )

    end = time.time()
    hybrid_times.append(end - start)

    hybrid_boxes_all.append(np.array(final_boxes))
    hybrid_scores_all.append(np.array(final_scores))
    hybrid_labels_all.append(np.array(final_labels))
    hybrid_det_counts.append(len(final_boxes))

avg_hybrid_time = np.mean(hybrid_times)
hybrid_fps = 1 / avg_hybrid_time

print(f" Hybrid Done!")
print(f"   Avg Time : {avg_hybrid_time:.3f}s")
print(f"   FPS      : {hybrid_fps:.1f}")
print(f"   Total Detections: {sum(hybrid_det_counts)}")
```

```python
def compute_simple_map(boxes_all, scores_all, iou_threshold=0.5):
    """Simple mAP calculation based on detections"""
    all_scores = []
    all_det_counts = []

    for boxes, scores in zip(boxes_all, scores_all):
        if len(scores) > 0:
            all_scores.extend(scores.tolist())
            all_det_counts.append(len(scores))
        else:
            all_det_counts.append(0)

    if len(all_scores) == 0:
        return 0.0

    avg_score = np.mean(all_scores)
    detection_rate = sum(1 for c in all_det_counts if c > 0) / len(all_det_counts)
    map_score = avg_score * detection_rate
    return round(map_score, 4)

yolo_map   = compute_simple_map(yolo_boxes_all,   yolo_scores_all)
frcnn_map  = compute_simple_map(frcnn_boxes_all,  frcnn_scores_all)
hybrid_map = compute_simple_map(hybrid_boxes_all, hybrid_scores_all)

print("=" * 40)
print(" mAP Results:")
print(f"   YOLO mAP@0.5        : {yolo_map:.4f}")
print(f"   Faster R-CNN mAP@0.5: {frcnn_map:.4f}")
print(f"   Hybrid mAP@0.5      : {hybrid_map:.4f}")
print("=" * 40)
```

```python
def compute_fp_fn(det_counts, expected=3):
    """Estimate FP and FN based on detection counts"""
    fps, fns = [], []
    for count in det_counts:
        fp = max(0, count - expected)
        fn = max(0, expected - count)
        fps.append(fp)
        fns.append(fn)
    return np.mean(fps), np.mean(fns)

yolo_fp,   yolo_fn   = compute_fp_fn(yolo_det_counts)
frcnn_fp,  frcnn_fn  = compute_fp_fn(frcnn_det_counts)
hybrid_fp, hybrid_fn = compute_fp_fn(hybrid_det_counts)

print(" False Positives / False Negatives:")
print(f"   YOLO        — FP: {yolo_fp:.2f}  FN: {yolo_fn:.2f}")
print(f"   Faster R-CNN— FP: {frcnn_fp:.2f}  FN: {frcnn_fn:.2f}")
print(f"   Hybrid      — FP: {hybrid_fp:.2f}  FN: {hybrid_fn:.2f}")
```

```python
speedup_vs_frcnn = avg_frcnn_time / avg_hybrid_time
speedup_vs_yolo  = avg_yolo_time  / avg_hybrid_time

print(" Speedup Analysis:")
print(f"   Hybrid vs Faster R-CNN : {speedup_vs_frcnn:.2f}x faster")
print(f"   Hybrid vs YOLO         : {speedup_vs_yolo:.2f}x")
print(f"   YOLO FPS               : {yolo_fps:.1f}")
print(f"   Faster R-CNN FPS       : {frcnn_fps:.1f}")
print(f"   Hybrid FPS             : {hybrid_fps:.1f}")
```

```python
COCO_CLASSES = [
    '__background__', 'person', 'bicycle', 'car', 'motorcycle',
    'airplane', 'bus', 'train', 'truck', 'boat', 'traffic light',
    'fire hydrant', 'stop sign', 'parking meter', 'bench', 'bird',
    'cat', 'dog', 'horse', 'sheep', 'cow', 'elephant', 'bear',
    'zebra', 'giraffe', 'backpack', 'umbrella', 'handbag', 'tie',
    'suitcase', 'frisbee', 'skis', 'snowboard', 'sports ball',
    'kite', 'baseball bat', 'baseball glove', 'skateboard',
    'surfboard', 'tennis racket', 'bottle', 'wine glass', 'cup',
    'fork', 'knife', 'spoon', 'bowl', 'banana', 'apple',
    'sandwich', 'orange', 'broccoli', 'carrot', 'hot dog', 'pizza',
    'donut', 'cake', 'chair', 'couch', 'potted plant', 'bed',
    'dining table', 'toilet', 'tv', 'laptop', 'mouse', 'remote',
    'keyboard', 'cell phone', 'microwave', 'oven', 'toaster',
    'sink', 'refrigerator', 'book', 'clock', 'vase', 'scissors',
    'teddy bear', 'hair drier', 'toothbrush'
]

def draw_boxes(image_rgb, boxes, labels=None,
               scores=None, color=(0,255,0), thickness=2):
    img = image_rgb.copy()
    for i, box in enumerate(boxes):
        x1, y1, x2, y2 = map(int, box)
        cv2.rectangle(img, (x1,y1), (x2,y2), color, thickness)
        if labels is not None and scores is not None:
            if i < len(labels) and i < len(scores):
                try:
                    lbl = COCO_CLASSES[int(labels[i])]
                except:
                    lbl = str(int(labels[i]))
                txt = f"{lbl} {float(scores[i]):.2f}"
                cv2.putText(img, txt, (x1, max(y1-5,10)),
                            cv2.FONT_HERSHEY_SIMPLEX,
                            0.5, color, 2)
    return img

# Pick sample image
sample_idx = 0
sample_img  = test_images[sample_idx]
sample_path = os.path.join(image_folder, sample_img)
orig = cv2.cvtColor(cv2.imread(sample_path), cv2.COLOR_BGR2RGB)

# Draw detections
img_yolo = draw_boxes(orig,
    yolo_boxes_all[sample_idx],
    yolo_labels_all[sample_idx],
    yolo_scores_all[sample_idx],
    color=(0,255,0))

img_frcnn = draw_boxes(orig,
    frcnn_boxes_all[sample_idx],
    frcnn_labels_all[sample_idx],
    frcnn_scores_all[sample_idx],
    color=(255,0,0))

img_hybrid = draw_boxes(orig,
    hybrid_boxes_all[sample_idx],
    hybrid_labels_all[sample_idx],
    hybrid_scores_all[sample_idx],
    color=(255,165,0))

# Plot all 3 side by side
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(img_yolo)
axes[0].set_title('Standalone YOLO', fontsize=13, fontweight='bold')
axes[0].axis('off')

axes[1].imshow(img_frcnn)
axes[1].set_title('Standalone Faster R-CNN', fontsize=13, fontweight='bold')
axes[1].axis('off')

axes[2].imshow(img_hybrid)
axes[2].set_title('Hybrid YOLO + Faster R-CNN', fontsize=13, fontweight='bold')
axes[2].axis('off')

plt.suptitle('Sample Detection Comparison', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('/content/plot7_sample_detections.png', dpi=150)
plt.show()
print(" Plot 7 saved!")
```
